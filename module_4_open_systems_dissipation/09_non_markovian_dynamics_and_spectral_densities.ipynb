### 1. Motivation and Physical Context
In standard open quantum system treatments (such as the Gorini-Kossakowski-Sudarshan-Lindblad master equation derived in Notebook 4.02), two fundamental approximations are invoked:
1. **Born Approximation:** Weak system-reservoir coupling ($g \ll \omega_0$), so that the reservoir state remains stationary $\rho_R(t) \approx \rho_R(0)$.
2. **Markov Approximation:** The bath correlation time $\tau_B = \hbar / \omega_c$ is infinitely shorter than the system relaxation timescale $\tau_R \sim 1/\gamma_0$. Consequently, reservoir memory decays instantaneously, giving memoryless generator dynamics $\dot{\rho}_S(t) = \mathcal{L}\rho_S(t)$.

However, in modern quantum technologies—including solid-state cavity QED (quantum dots in micropillars), superconducting circuit QED coupled to transmission lines, photosynthetic pigment-protein complexes, and photonic crystal bandgaps—these approximations break down dramatically:
- The reservoir is **structured** (possessing band edges, narrow resonances, or photonic band gaps where density of states drops to zero or exhibits singularities).
- The bath correlation time $\tau_B$ becomes comparable to or exceeds the system evolution timescale $\tau_R$.
- **Information backflow:** Information previously lost to environmental degrees of freedom temporarily returns to the system, leading to non-exponential decay, population oscillations, and revivals of quantum coherence and entanglement.

---

### 2. Microscopic Hamiltonian & Exact Integro-Differential Dynamics
Consider a two-level system (atom or qubit) with transition frequency $\omega_0$ coupled linearly to a bosonic reservoir at zero temperature ($T = 0$):
$$\hat{H} = \hbar \omega_0 \hat{\sigma}_+ \hat{\sigma}_- + \sum_k \hbar \omega_k \hat{b}_k^\dagger \hat{b}_k + \sum_k \hbar \left( g_k \hat{\sigma}_+ \hat{b}_k + g_k^* \hat{\sigma}_- \hat{b}_k^\dagger \right)$$

Working in the interaction picture and restricting to the single-excitation subspace:
$$|\psi(t)\rangle = c_e(t) |e, 0\rangle + \sum_k c_k(t) |g, 1_k\rangle$$

The Schrödinger equation yields the coupled differential equations:
$$i \dot{c}_e(t) = \sum_k g_k c_k(t) e^{i(\omega_0 - \omega_k)t}, \qquad i \dot{c}_k(t) = g_k^* c_e(t) e^{-i(\omega_0 - \omega_k)t}$$

Integrating the equation for $c_k(t)$ subject to the vacuum initial condition $c_k(0) = 0$ and substituting back into the equation for $c_e(t)$ yields the **exact Volterra integro-differential equation**:
$$\dot{c}_e(t) = -\int_0^t f(t - t') c_e(t') \, dt'$$
where the environmental **memory kernel** (bath correlation function) $f(\tau)$ is the Fourier transform of the environmental **spectral density** $J(\omega) = \sum_k |g_k|^2 \delta(\omega - \omega_k)$:
$$f(\tau) = \int_0^\infty J(\omega) e^{-i(\omega - \omega_0)\tau} \, d\omega$$

---

### 3. Spectral Densities: Continuous Bath Structuring
The spectral density $J(\omega)$ encodes both the environmental density of states $g(\omega)$ and the frequency-dependent coupling strength:
1. **Power-Law Reservoirs (Ohmic, Sub-Ohmic, Super-Ohmic):**
   $$J(\omega) = \alpha \omega \left(\frac{\omega}{\omega_c}\right)^{s-1} e^{-\omega/\omega_c}$$
   - $s = 1$: **Ohmic** bath (linear frequency dependence, typical of 1D waveguides and electromagnetic fields in free space).
   - $s < 1$: **Sub-Ohmic** bath (divergent low-frequency coupling, dominant in quantum dots with acoustic phonon coupling).
   - $s > 1$: **Super-Ohmic** bath (suppressed low frequencies, characteristic of 3D acoustic phonons with $s=3$).

2. **Lorentzian (Cavity-Structured) Reservoir:**
   When an atom interacts with a leaky single-mode cavity of frequency $\omega_c$, decay width $\lambda$, and atom-field coupling $g$:
   $$J(\omega) = \frac{1}{2\pi} \frac{\gamma_0 \lambda^2}{(\omega - \omega_c)^2 + \lambda^2}$$
   Here $\lambda$ is the spectral bandwidth ($\tau_B \sim 1/\lambda$), and $\gamma_0$ is the Markovian decay rate.
   On resonance ($\omega_c = \omega_0$), the memory kernel is purely exponential:
   $$f(\tau) = \frac{\gamma_0 \lambda}{2} e^{-\lambda \tau}$$

---

### 4. Exact Analytical Solution for Lorentzian Bath
Applying the Laplace transform $\mathcal{L}\{c_e(t)\} = \tilde{c}_e(s)$:
$$s \tilde{c}_e(s) - 1 = -\tilde{f}(s) \tilde{c}_e(s) \implies \tilde{c}_e(s) = \frac{1}{s + \frac{\gamma_0 \lambda / 2}{s + \lambda}} = \frac{s + \lambda}{s^2 + \lambda s + \frac{1}{2}\gamma_0 \lambda}$$

The poles are given by the roots of the denominator:
$$s_{1,2} = -\frac{\lambda}{2} \pm \frac{1}{2} \sqrt{\lambda^2 - 2\gamma_0 \lambda} \equiv -\frac{\lambda}{2} \pm \frac{d}{2}$$

We identify two sharply distinct dynamical regimes:
1. **Weak Coupling / Markovian Regime ($\lambda > 2\gamma_0$):**
   $d = \sqrt{\lambda^2 - 2\gamma_0 \lambda} \in \mathbb{R}$. Both poles are real and negative.
   $$c_e(t) = e^{-\lambda t / 2} \left( \cosh\frac{d t}{2} + \frac{\lambda}{d} \sinh\frac{d t}{2} \right)$$
   The population $P_e(t) = |c_e(t)|^2$ decays monotonically without any revivals. In the broadband limit $\lambda \gg \gamma_0$, $d \approx \lambda - \gamma_0$, recovering standard exponential Lindblad decay $c_e(t) \approx e^{-\gamma_0 t / 2}$.

2. **Strong Coupling / Non-Markovian Regime ($\lambda < 2\gamma_0$):**
   $d = i \Omega$ is purely imaginary, with generalized Rabi frequency $\Omega = \sqrt{2\gamma_0 \lambda - \lambda^2}$.
   $$c_e(t) = e^{-\lambda t / 2} \left( \cos\frac{\Omega t}{2} + \frac{\lambda}{\Omega} \sin\frac{\Omega t}{2} \right)$$
   The excited state population $P_e(t) = |c_e(t)|^2$ exhibits damped oscillations (vacuum Rabi revivals), reflecting reversible exchange of virtual photons between the atom and the structured cavity reservoir.

---

### 5. Time-Convolutionless (TCL) Master Equation & Negative Decay Rates
The exact reduced density matrix $\rho_S(t)$ can be cast into the canonical Time-Convolutionless form:
$$\frac{d}{dt}\rho_S(t) = -i [S(t) \hat{\sigma}_+ \hat{\sigma}_-, \rho_S(t)] + \gamma(t) \left( \hat{\sigma}_- \rho_S(t) \hat{\sigma}_+ - \frac{1}{2}\{\hat{\sigma}_+ \hat{\sigma}_-, \rho_S(t)\} \right)$$
where the canonical, time-dependent decay rate is given by:
$$\gamma(t) = -2 \, \text{Re}\left( \frac{\dot{c}_e(t)}{c_e(t)} \right) = -\frac{1}{P_e(t)} \frac{d P_e(t)}{dt}$$

> **Key Physical Criterion:**
> In a Markovian semigroup, $\gamma(t) = \gamma_0 \geq 0$ for all $t$.
> Whenever $\gamma(t) < 0$, the generator is temporarily non-positive, which corresponds directly to $\frac{d P_e(t)}{dt} > 0$: **population and information flow back into the system from the environment**.

---

### 6. Breuer-Laine-Piilo (BLP) Non-Markovianity Measure
Breuer, Laine, and Piilo (PRL 103, 210401 (2009)) introduced an operational, witness-free measure of non-Markovianity based on the **trace distance** $D(\rho_1, \rho_2) = \frac{1}{2} \text{Tr}|\rho_1 - \rho_2|$ between two quantum states:
- For any completely positive and trace-preserving (CPTP) Markovian semigroup, trace distance is strictly contractive: $\sigma(t) = \frac{d}{dt} D(\rho_1(t), \rho_2(t)) \leq 0$.
- Any growth in trace distance ($\sigma(t) > 0$) signifies non-Markovian information backflow.
- The total non-Markovianity measure is defined as:
  $$\mathcal{N}_{\text{BLP}} = \max_{\rho_{1,2}(0)} \int_{\sigma > 0} \sigma(t) \, dt$$
  For our resonant two-level system with optimal initial states $|e\rangle$ and $|g\rangle$, $D(t) = |c_e(t)|^2 = P_e(t)$, yielding:
  $$\sigma(t) = \frac{d P_e(t)}{dt}, \qquad \mathcal{N}_{\text{BLP}} = \int_{\dot{P}_e > 0} \frac{d P_e(t)}{dt} \, dt = \sum_k \left[ P_e(t_{\text{max}, k}) - P_e(t_{\text{min}, k}) \right]$$
  - If $\lambda \geq 2\gamma_0$: $P_e(t)$ decays monotonically $\implies \mathcal{N}_{\text{BLP}} \equiv 0$ (Markovian).
  - If $\lambda < 2\gamma_0$: $P_e(t)$ oscillates $\implies \mathcal{N}_{\text{BLP}} > 0$ (Strictly Non-Markovian).

```python
class NonMarkovianTwoLevelEngine:
    r"""
    Comprehensive computational engine for non-Markovian open quantum system dynamics.
    
    Provides:
      1. Exact analytical dynamics for a two-level emitter coupled to a resonant Lorentzian reservoir.
      2. High-precision auxiliary-mode / Volterra memory solver via embedded 4th-order Runge-Kutta.
      3. Canonical time-dependent decay rate gamma(t) computation.
      4. Breuer-Laine-Piilo (BLP) trace distance information backflow measure.
      5. Parametric spectral density generators (Ohmic, Sub-Ohmic, Super-Ohmic, and Lorentzian).
    """
    def __init__(self, gamma0=1.0, omega0=10.0):
        self.gamma0 = float(gamma0)
        self.omega0 = float(omega0)
        
    def analytical_amplitude(self, t_arr, lambda_param):
        r"""
        Computes analytical probability amplitude c_e(t) for resonant Lorentzian reservoir.
        """
        t = np.asarray(t_arr, dtype=float)
        gamma0 = self.gamma0
        lam = float(lambda_param)
        
        disc = lam**2 - 2.0 * gamma0 * lam
        
        if np.isclose(disc, 0.0, atol=1e-12):
            # Critical damping: lambda = 2*gamma0
            c_e = (1.0 + 0.5 * lam * t) * np.exp(-0.5 * lam * t)
        elif disc > 0.0:
            # Overdamped / Markovian-like monotonic decay: lambda > 2*gamma0
            d = np.sqrt(disc)
            c_e = np.exp(-0.5 * lam * t) * (np.cosh(0.5 * d * t) + (lam / d) * np.sinh(0.5 * d * t))
        else:
            # Underdamped / Non-Markovian oscillations: lambda < 2*gamma0
            Omega = np.sqrt(-disc)
            c_e = np.exp(-0.5 * lam * t) * (np.cos(0.5 * Omega * t) + (lam / Omega) * np.sin(0.5 * Omega * t))
            
        return c_e
    
    def analytical_population(self, t_arr, lambda_param):
        r"""
        Computes excited state population P_e(t) = |c_e(t)|^2.
        """
        c_e = self.analytical_amplitude(t_arr, lambda_param)
        return np.abs(c_e)**2
    
    def numerical_volterra_solver(self, t_arr, lambda_param):
        r"""
        Solves dc_e/dt = - \int_0^t f(t-t') c_e(t') dt' using the exact auxiliary-mode
        state-space transformation:
            dc_e/dt = -z(t)
            dz/dt   = (1/2)*gamma0*lambda*c_e(t) - lambda*z(t)
        integrated with 4th-order Runge-Kutta (RK4).
        """
        t = np.asarray(t_arr, dtype=float)
        dt = t[1] - t[0]
        N = len(t)
        lam = float(lambda_param)
        gamma0 = self.gamma0
        f0 = 0.5 * gamma0 * lam
        
        c_e = np.zeros(N, dtype=float)
        z = np.zeros(N, dtype=float)
        c_e[0] = 1.0
        z[0] = 0.0
        
        for n in range(N - 1):
            def deriv(c_val, z_val):
                return -z_val, f0 * c_val - lam * z_val
            
            k1_c, k1_z = deriv(c_e[n], z[n])
            k2_c, k2_z = deriv(c_e[n] + 0.5 * dt * k1_c, z[n] + 0.5 * dt * k1_z)
            k3_c, k3_z = deriv(c_e[n] + 0.5 * dt * k2_c, z[n] + 0.5 * dt * k2_z)
            k4_c, k4_z = deriv(c_e[n] + dt * k3_c, z[n] + dt * k3_z)
            
            c_e[n + 1] = c_e[n] + (dt / 6.0) * (k1_c + 2.0 * k2_c + 2.0 * k3_c + k4_c)
            z[n + 1]   = z[n]   + (dt / 6.0) * (k1_z + 2.0 * k2_z + 2.0 * k3_z + k4_z)
            
        return c_e
    
    def canonical_decay_rate(self, t_arr, lambda_param):
        r"""
        Computes canonical time-dependent decay rate gamma(t) = - (1/P_e) dP_e/dt.
        """
        t = np.asarray(t_arr, dtype=float)
        P_e = self.analytical_population(t, lambda_param)
        dP_dt = np.gradient(P_e, t)
        
        safe_Pe = np.maximum(P_e, 1e-12)
        gamma_t = - dP_dt / safe_Pe
        gamma_t[P_e < 1e-6] = 0.0
        return gamma_t
    
    def blp_measure(self, t_arr, lambda_param):
        r"""
        Computes Breuer-Laine-Piilo non-Markovianity measure N_BLP = \int_{sigma > 0} sigma(t) dt.
        For resonant two-level system with initial states |e> and |g>, sigma(t) = dP_e/dt.
        """
        t = np.asarray(t_arr, dtype=float)
        P_e = self.analytical_population(t, lambda_param)
        dP_dt = np.gradient(P_e, t)
        
        sigma_pos = np.maximum(dP_dt, 0.0)
        n_blp = np.trapezoid(sigma_pos, t) if hasattr(np, 'trapezoid') else np.trapz(sigma_pos, t)
        return float(n_blp)
    
    @staticmethod
    def spectral_density_powerlaw(omega, alpha=0.1, omega_c=5.0, s=1.0):
        r"""
        Power-law spectral density J(omega) = alpha * omega * (omega/omega_c)**(s-1) * exp(-omega/omega_c).
        """
        w = np.asarray(omega, dtype=float)
        return alpha * w * ((w / omega_c)**(s - 1.0)) * np.exp(-w / omega_c)

    @staticmethod
    def spectral_density_lorentzian(omega, gamma0=1.0, lambda_param=0.5, omega_0=5.0):
        r"""
        Lorentzian resonant spectral density J(omega) = (1/2pi) * (gamma0 * lambda**2) / ((omega - omega_0)**2 + lambda**2).
        """
        w = np.asarray(omega, dtype=float)
        return (1.0 / (2.0 * np.pi)) * (gamma0 * lambda_param**2) / ((w - omega_0)**2 + lambda_param**2)

print("NonMarkovianTwoLevelEngine successfully defined.")

```

```python
# Instantiate the computational engine
engine = NonMarkovianTwoLevelEngine(gamma0=1.0, omega0=10.0)

# Time axis: 0 to 18 / gamma0
t = np.linspace(0.0, 18.0, 1000)

# Parameter exploration: Bandwidth lambda / gamma0
lambda_values = [0.1, 0.4, 1.0, 2.0, 5.0]
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

fig = plt.figure(figsize=(16, 12), dpi=300)
gs = GridSpec(2, 2, figure=fig, hspace=0.28, wspace=0.24)

# ------------------------------------------------------------------------------
# PANEL (A): Excited State Population P_e(t) (Markovian vs Non-Markovian)
# ------------------------------------------------------------------------------
ax_a = fig.add_subplot(gs[0, 0])

for lam, col in zip(lambda_values, colors):
    P_e = engine.analytical_population(t, lam)
    label = rf"$\lambda = {lam}\,\gamma_0$"
    if lam < 2.0:
        label += r" (Non-Markovian)"
    elif np.isclose(lam, 2.0):
        label += r" (Critical $\lambda_c$)"
    else:
        label += r" (Markovian)"
    ax_a.plot(t, P_e, label=label, color=col, linewidth=2.2)

# Markovian exponential decay limit
ax_a.plot(t, np.exp(-engine.gamma0 * t), 'k--', linewidth=1.8, label=r"Pure Markovian $e^{-\gamma_0 t}$", alpha=0.85)

ax_a.set_title(r"$\mathbf{(a)}$ Excited State Population Dynamics $P_e(t)$", pad=10)
ax_a.set_xlabel(r"Dimensionless Time $\gamma_0 t$")
ax_a.set_ylabel(r"Excited Population $P_e(t)$")
ax_a.set_xlim(0, 18)
ax_a.set_ylim(-0.02, 1.02)
ax_a.legend(loc="upper right", framealpha=0.92)
ax_a.grid(True)

# ------------------------------------------------------------------------------
# PANEL (B): Canonical Time-Dependent Decay Rate gamma(t) & Negative Regions
# ------------------------------------------------------------------------------
ax_b = fig.add_subplot(gs[0, 1])

t_rate = np.linspace(0.05, 12.0, 800)
lam_nm = 0.2
gamma_nm = engine.canonical_decay_rate(t_rate, lam_nm)
P_nm = engine.analytical_population(t_rate, lam_nm)

# Mask negative and positive rates
gamma_neg = np.where(gamma_nm < 0.0, gamma_nm, np.nan)
gamma_pos = np.where(gamma_nm >= 0.0, gamma_nm, np.nan)

ax_b.plot(t_rate, gamma_pos, color='#1f77b4', linewidth=2.2, label=rf"$\gamma(t) \geq 0$ ($\lambda = {lam_nm}\,\gamma_0$)")
ax_b.plot(t_rate, gamma_neg, color='#d62728', linewidth=2.5, label=rf"$\gamma(t) < 0$ (Information Backflow)")

# Fill negative region
ax_b.fill_between(t_rate, 0, np.nan_to_num(gamma_neg, nan=0.0), color='#d62728', alpha=0.25, label=r"Backflow Domain $\gamma(t) < 0$")

# Plot a Markovian case for comparison
gamma_markov = engine.canonical_decay_rate(t_rate, 4.0)
ax_b.plot(t_rate, gamma_markov, color='#2ca02c', linestyle='-.', linewidth=2.0, label=r"Markovian $\lambda = 4.0\,\gamma_0$")

ax_b.axhline(0, color='black', linestyle=':', linewidth=1.2)
ax_b.axhline(engine.gamma0, color='gray', linestyle='--', linewidth=1.2, label=r"Standard Rate $\gamma_0$")

ax_b.set_title(r"$\mathbf{(b)}$ Canonical Time-Dependent Decay Rate $\gamma(t)$", pad=10)
ax_b.set_xlabel(r"Dimensionless Time $\gamma_0 t$")
ax_b.set_ylabel(r"Decay Rate $\gamma(t) / \gamma_0$")
ax_b.set_xlim(0, 12)
ax_b.set_ylim(-1.5, 2.5)
ax_b.legend(loc="upper right", framealpha=0.92)
ax_b.grid(True)

# ------------------------------------------------------------------------------
# PANEL (C): Trace Distance Derivative and Cumulative BLP Backflow
# ------------------------------------------------------------------------------
ax_c = fig.add_subplot(gs[1, 0])

for lam, col in zip([0.1, 0.3, 0.8], ['#1f77b4', '#ff7f0e', '#2ca02c']):
    P_e_curve = engine.analytical_population(t, lam)
    sigma_t = np.gradient(P_e_curve, t)
    sigma_pos = np.maximum(sigma_t, 0.0)
    cum_backflow = integrate.cumulative_trapezoid(sigma_pos, t, initial=0.0)
    
    ax_c.plot(t, cum_backflow, label=rf"$\mathcal{{N}}(t)$ for $\lambda = {lam}\,\gamma_0$", color=col, linewidth=2.2)

ax_c.set_title(r"$\mathbf{(c)}$ Cumulative BLP Information Backflow $\mathcal{N}(t)$", pad=10)
ax_c.set_xlabel(r"Dimensionless Time $\gamma_0 t$")
ax_c.set_ylabel(r"Cumulative Backflow $\int_0^t \max(\sigma(t'), 0) dt'$")
ax_c.set_xlim(0, 18)
ax_c.set_ylim(-0.02, 0.8)
ax_c.legend(loc="upper left", framealpha=0.92)
ax_c.grid(True)

# ------------------------------------------------------------------------------
# PANEL (D): Spectral Densities and Non-Markovianity Phase Diagram
# ------------------------------------------------------------------------------
ax_d = fig.add_subplot(gs[1, 1])

omega = np.linspace(0.01, 15.0, 500)
J_sub = engine.spectral_density_powerlaw(omega, alpha=0.3, omega_c=4.0, s=0.5)
J_ohm = engine.spectral_density_powerlaw(omega, alpha=0.3, omega_c=4.0, s=1.0)
J_sup = engine.spectral_density_powerlaw(omega, alpha=0.3, omega_c=4.0, s=2.5)
J_lor = engine.spectral_density_lorentzian(omega, gamma0=2.0, lambda_param=1.0, omega_0=6.0)

ax_d.plot(omega, J_sub, label=r"Sub-Ohmic ($s=0.5$)", color='#ff7f0e', linewidth=2.0)
ax_d.plot(omega, J_ohm, label=r"Ohmic ($s=1.0$)", color='#1f77b4', linewidth=2.0)
ax_d.plot(omega, J_sup, label=r"Super-Ohmic ($s=2.5$)", color='#2ca02c', linewidth=2.0)
ax_d.plot(omega, J_lor, label=r"Lorentzian Cavity Peak", color='#d62728', linewidth=2.2, linestyle='--')

ax_d.set_title(r"$\mathbf{(d)}$ Environmental Spectral Densities $J(\omega)$", pad=10)
ax_d.set_xlabel(r"Frequency $\omega / \gamma_0$")
ax_d.set_ylabel(r"Spectral Coupling $J(\omega)$")
ax_d.set_xlim(0, 15)
ax_d.set_ylim(0, 0.7)
ax_d.legend(loc="upper right", framealpha=0.92)
ax_d.grid(True)

# Inset: BLP Measure N_BLP vs Reservoir Bandwidth lambda
inset_ax = ax_d.inset_axes([0.48, 0.22, 0.48, 0.38])
lam_scan = np.linspace(0.05, 3.0, 60)
n_blp_scan = [engine.blp_measure(t, l_val) for l_val in lam_scan]

inset_ax.plot(lam_scan, n_blp_scan, color='#9467bd', linewidth=2.2)
inset_ax.axvline(2.0, color='red', linestyle=':', label=r"$\lambda_c = 2\gamma_0$")
inset_ax.set_xlabel(r"$\lambda / \gamma_0$", fontsize=9)
inset_ax.set_ylabel(r"$\mathcal{N}_{\mathrm{BLP}}$", fontsize=9)
inset_ax.set_title(r"Non-Markovianity Phase Transition", fontsize=9)
inset_ax.tick_params(labelsize=8)
inset_ax.grid(True, alpha=0.3)
inset_ax.legend(fontsize=8, loc='upper right')

output_fig_path = "non_markovian_spectral_densities.png"
fig.savefig(output_fig_path, dpi=300, bbox_inches='tight')
plt.close(fig)

print(f"Publication-quality 4-panel figure successfully saved to '{output_fig_path}'.")

```

### 7. Physical Interpretation & Key Findings

1. **The Sharp Non-Markovian Boundary at $\lambda = 2\gamma_0$:**
   In a Lorentzian reservoir, the ratio of reservoir bandwidth to coupling strength, $\lambda / \gamma_0$, governs the dynamical regime:
   - When $\lambda > 2\gamma_0$, the bath memory decays faster than the atom can absorb virtual excitations. The eigenvalues $s_{1,2}$ are purely real, resulting in overdamped, strictly monotonic decay ($P_e(t)$ monotonically decreasing).
   - At $\lambda_c = 2\gamma_0$, critical damping occurs.
   - When $\lambda < 2\gamma_0$, the roots develop an imaginary part $d = i\Omega = i\sqrt{2\gamma_0 \lambda - \lambda^2}$. The emitter undergoes coherent vacuum Rabi oscillations with its own localized cavity reservoir mode, manifesting damped revivals of population.

2. **Negative Canonical Decay Rates as Information Backflow:**
   Panel (b) reveals that during periods where population revives ($\dot{P}_e(t) > 0$), the canonical Time-Convolutionless decay rate $\gamma(t)$ dives into negative values ($\gamma(t) < 0$). In the operator picture, a negative Lindblad rate $\gamma(t) < 0$ acts as a temporary amplifier/purifier rather than a dissipator, restoring quantum information to the two-level emitter.

3. **Continuous Family of Spectral Densities:**
   Panel (d) illustrates how structural modifications of the environment alter open-system behavior:
   - **Sub-Ohmic ($s < 1$):** Enhanced low-frequency modes cause severe dephasing and power-law long-time algebraic tails.
   - **Ohmic ($s = 1$):** Standard linear low-frequency profile, typical of resistor noise (Johnson-Nyquist) and electromagnetic field free space.
   - **Super-Ohmic ($s > 1$):** Low frequencies are suppressed; coupling peaks at intermediate energy before the exponential cutoff.
   - The inset confirms that the operational BLP measure $\mathcal{N}_{\text{BLP}}$ drops smoothly to zero exactly at $\lambda = 2\gamma_0$, validating the mathematical boundary of the Markovian semigroup.
