## Theoretical Framework: Quantum Laser Theory & The Micromaser

### 1. Semiclassical vs. Quantized Laser Theory
In semiclassical laser theory (Lamb theory), the electromagnetic field is treated as a classical wave $E(t)$ obeying Maxwell's equations driven by quantum expectation values of atomic dipoles. 
While semiclassical theory predicts the threshold condition and operating frequency, it fails completely to describe:
* Laser linewidth (Schawlow-Townes limit).
* Spontaneous emission noise initiating lasing.
* Quantum fluctuations and photon statistics $P(n)$.

The **Scully-Lamb master equation** (1967) treats both the field and gain medium fully quantum mechanically.

---

### 2. The Scully-Lamb Quantum Laser Master Equation
Consider a single cavity mode $\hat{a}$ with damping rate $\kappa$, pumped by inverted two-level atoms injected at rate $r$. 
Adiabatically eliminating the fast atomic polarizations and populations yields the master equation for the diagonal elements of the cavity density matrix $p_n \equiv \rho_{nn} = \langle n|\hat{\rho}|n\rangle$:

$$\frac{dp_n}{dt} = -\frac{A(n+1)}{1 + (B/A)(n+1)} p_n + \frac{An}{1 + (B/A)n} p_{n-1} - \kappa n p_n + \kappa (n+1) p_{n+1}$$

where:
* $A$ is the **linear gain coefficient** (stimulated emission into mode): $A = r \frac{2g^2}{\gamma_{\perp}^2}$.
* $B$ is the **self-saturation coefficient**: $B = r \frac{4g^4}{\gamma_{\perp}^4}$.
* $\kappa$ is the **cavity loss rate**.

In steady state ($\frac{dp_n}{dt} = 0$), detailed balance between adjacent Fock levels requires:
$$p_n = p_{n-1} \frac{A / \kappa}{1 + (B/A) n} = p_0 \prod_{k=1}^n \frac{A / \kappa}{1 + (B/A) k}$$
where $p_0$ is fixed by total probability normalization $\sum_{n=0}^{\infty} p_n = 1$.

#### Physical Regimes:
1. **Below Threshold ($A < \kappa$):** Linear gain cannot overcome cavity losses. For small $n$, $p_n \approx p_0 (A/\kappa)^n$, which is a **Bose-Einstein thermal geometric distribution** with average photon number $\langle n\rangle = \frac{A}{\kappa - A}$ and second-order intensity correlation $g^{(2)}(0) = 2$.
2. **At Threshold ($A = \kappa$):** The system undergoes a non-equilibrium continuous (second-order) phase transition characterized by large, critical macroscopic fluctuations.
3. **Above Threshold ($A > \kappa$):** Saturation stabilizes the intensity at $\bar{n} \approx \frac{A - \kappa}{B}$. The photon distribution sharpens into a narrow, **Poissonian coherent distribution** with $g^{(2)}(0) \to 1$ and Mandel parameter $Q \to 0$.

---

### 3. The Micromaser (One-Atom Maser)
In a conventional laser, thousands of atoms interact simultaneously with the cavity field. 
In the **micromaser** (Meschede, Walther, and Müller 1985; Haroche 1990s):
* Ultra-cold excited Rydberg atoms are injected into a superconducting niobium microwave cavity ($Q \sim 10^{10}$) at rate $R$.
* The atomic flux is so dilute that at most **one single atom** is inside the cavity at any given moment.
* Each atom spends an identical, precisely controlled interaction transit time $\tau_{\text{int}}$ inside the cavity.

The steady-state master equation (Filipovicz, Javanainen, and Meystre 1986) is governed by detailed balance:
$$p_n = p_{n-1} \frac{N_{\text{ex}}}{n} \sin^2\left(g \tau_{\text{int}} \sqrt{n}\right) = p_0 \prod_{k=1}^n \frac{N_{\text{ex}}}{k} \sin^2\left(g \tau_{\text{int}} \sqrt{k}\right)$$
where $N_{\text{ex}} \equiv R / \kappa$ is the average number of atoms entering the cavity during the cavity photon decay lifetime. 
The dimensionless **pump parameter** is defined as:
$$\theta \equiv g \tau_{\text{int}} \sqrt{N_{\text{ex}}}$$

---

### 4. Quantum Trapping States & Sub-Poissonian Light
The probability that an atom entering in excited state $|e\rangle$ emits a photon into a cavity containing $n$ photons is:
$$P_{e \to g}(n) = \sin^2\left(g \tau_{\text{int}} \sqrt{n+1}\right)$$

If there exists an integer $n_q$ such that:
$$g \tau_{\text{int}} \sqrt{n_q + 1} = q \pi \quad (q = 1, 2, 3, \dots)$$
then:
$$\sin^2\left(g \tau_{\text{int}} \sqrt{n_q + 1}\right) = \sin^2(q\pi) \equiv 0!$$

#### The Trapping Phenomenon:
1. When the cavity has $n_q$ photons, an entering atom undergoes an exact $2q\pi$ Rabi flop and emerges in $|e\rangle$ **without emitting a photon**!
2. Upward transitions from $n_q \to n_q + 1$ are **completely blocked**!
3. Consequently, the photon distribution is strictly truncated:
   $$p_n \equiv 0 \quad \forall n > n_q$$
This creates a **quantum trapping state**: the cavity field is macroscopically confined within a finite Fock subspace, generating extreme sub-Poissonian non-classical light with negative Mandel parameter $Q < 0$ and near-deterministic Fock states ($Q \to -1$).



```python
# Implementation: Scully-Lamb Laser & Micromaser Simulation Engine

class QuantumLaserEngine:
    """
    Simulation engine for Scully-Lamb quantum laser theory and micromaser trapping states.
    """
    def __init__(self, n_cutoff=100):
        self.n_cutoff = int(n_cutoff)
        self.n_vals = np.arange(self.n_cutoff, dtype=float)

    def scully_lamb_distribution(self, A=1.0, B=0.02, kappa=1.0):
        """
        Compute the Scully-Lamb steady-state photon number distribution P(n):
            p_n = p_{n-1} * (A / kappa) / (1 + (B/A)*n)
        """
        A = float(A)
        B = float(B)
        kappa = float(kappa)
        
        log_p = np.zeros(self.n_cutoff, dtype=float)
        sat_ratio = B / A
        gain_loss_ratio = A / kappa
        
        for n in range(1, self.n_cutoff):
            # Use log-domain accumulation to prevent numerical underflow/overflow
            step_factor = gain_loss_ratio / (1.0 + sat_ratio * n)
            log_p[n] = log_p[n-1] + np.log(max(step_factor, 1e-300))
            
        # Shift log probabilities to prevent float overflow before exp
        log_p -= np.max(log_p)
        p = np.exp(log_p)
        p /= np.sum(p)  # Strict normalization
        
        return p

    def micromaser_distribution(self, N_ex=50.0, g_tau=0.25):
        """
        Compute the steady-state photon distribution of a micromaser:
            p_n = p_{n-1} * (N_ex / n) * sin^2(g * tau * sqrt(n))
        """
        N_ex = float(N_ex)
        g_tau = float(g_tau)
        
        log_p = np.zeros(self.n_cutoff, dtype=float)
        
        for n in range(1, self.n_cutoff):
            sin_term = np.sin(g_tau * np.sqrt(n))**2
            if sin_term < 1e-15:
                # Trapping state cutoff reached: upward transitions blocked
                log_p[n:] = -1000.0  # Zero probability for all higher n
                break
            else:
                step = (N_ex / n) * sin_term
                log_p[n] = log_p[n-1] + np.log(max(step, 1e-300))
                
        log_p -= np.max(log_p)
        p = np.exp(log_p)
        p /= np.sum(p)
        
        return p

    def compute_statistics(self, p_dist):
        """
        Compute mean photon number <n>, variance Var(n), Mandel Q parameter, and g^(2)(0).
        """
        n = self.n_vals[:len(p_dist)]
        mean_n = float(np.sum(n * p_dist))
        mean_n2 = float(np.sum((n**2) * p_dist))
        var_n = mean_n2 - mean_n**2
        
        # Mandel Q = (Var(n) - <n>) / <n>
        Q = (var_n - mean_n) / mean_n if mean_n > 1e-12 else 0.0
        
        # g^(2)(0) = <n(n-1)> / <n>^2 = 1 + Q / <n>
        g2_0 = (mean_n2 - mean_n) / (mean_n**2) if mean_n > 1e-12 else 2.0
        
        return {
            "mean_n": mean_n,
            "var_n": var_n,
            "mandel_Q": Q,
            "g2_0": g2_0
        }

print("QuantumLaserEngine successfully loaded.")

```

```python
# Publication-Quality 4-Panel Visualization: Laser Phase Transition & Micromaser Trapping States

engine = QuantumLaserEngine(n_cutoff=120)

# 1. Scully-Lamb distributions across threshold (A/kappa = 0.8, 1.0, 1.5, 3.0)
kappa_ref = 1.0
B_ref = 0.04
p_sub = engine.scully_lamb_distribution(A=0.8, B=B_ref, kappa=kappa_ref)
p_thresh = engine.scully_lamb_distribution(A=1.0, B=B_ref, kappa=kappa_ref)
p_above1 = engine.scully_lamb_distribution(A=1.8, B=B_ref, kappa=kappa_ref)
p_above2 = engine.scully_lamb_distribution(A=3.2, B=B_ref, kappa=kappa_ref)

# 2. Laser threshold curve: <n> and g^(2)(0) vs. A/kappa
pump_ratios = np.linspace(0.4, 3.5, 80)
mean_n_curve = []
g2_curve = []
Q_curve = []

for r in pump_ratios:
    p_r = engine.scully_lamb_distribution(A=r, B=B_ref, kappa=kappa_ref)
    stats_r = engine.compute_statistics(p_r)
    mean_n_curve.append(stats_r["mean_n"])
    g2_curve.append(stats_r["g2_0"])
    Q_curve.append(stats_r["mandel_Q"])

mean_n_curve = np.array(mean_n_curve)
g2_curve = np.array(g2_curve)

# 3. Micromaser trapping states:
# Condition: g * tau * sqrt(n_q + 1) = pi
# Choose n_q = 15: g_tau = pi / sqrt(16) = pi / 4 = 0.785398
nq_target = 15
g_tau_trap = np.pi / np.sqrt(nq_target + 1)
p_trap = engine.micromaser_distribution(N_ex=60.0, g_tau=g_tau_trap)
p_notrap = engine.micromaser_distribution(N_ex=60.0, g_tau=0.45)

# 4. Micromaser Mandel Q vs. pump parameter theta = g_tau * sqrt(N_ex)
g_tau_scan = np.linspace(0.1, 1.8, 120)
Nex_fixed = 50.0
theta_vals = g_tau_scan * np.sqrt(Nex_fixed)
Q_micro_curve = []

for gt in g_tau_scan:
    p_gt = engine.micromaser_distribution(N_ex=Nex_fixed, g_tau=gt)
    st = engine.compute_statistics(p_gt)
    Q_micro_curve.append(st["mandel_Q"])

# Generate 4-Panel Masterpiece Figure
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# -------------------------------------------------------------
# Panel (a): Scully-Lamb Photon Distributions P(n)
# -------------------------------------------------------------
ax_a = axes[0, 0]
n_axis = np.arange(80)
ax_a.plot(n_axis, p_sub[:80], "o-", color="#3498db", markersize=4, label=r"Below Threshold ($A/\kappa = 0.8$, Thermal)")
ax_a.plot(n_axis, p_thresh[:80], "s-", color="#f39c12", markersize=4, label=r"At Threshold ($A/\kappa = 1.0$)")
ax_a.plot(n_axis, p_above1[:80], "^-", color="#2ecc71", markersize=4, label=r"Above ($A/\kappa = 1.8$, Coherent)")
ax_a.plot(n_axis, p_above2[:80], "d-", color="#e74c3c", markersize=4, label=r"Far Above ($A/\kappa = 3.2$, Coherent)")

ax_a.set_title(r"(a) Scully-Lamb Photon Statistics Across Threshold", fontweight="bold")
ax_a.set_xlabel(r"Photon Number $n$")
ax_a.set_ylabel(r"Probability $P(n)$")
ax_a.set_xlim(-1, 75)
ax_a.set_ylim(-0.005, 0.16)
ax_a.legend(loc="upper right", framealpha=0.9)

# -------------------------------------------------------------
# Panel (b): Laser Threshold Transition: <n> and g^(2)(0)
# -------------------------------------------------------------
ax_b = axes[0, 1]
color_n = "#2c3e50"
color_g2 = "#e74c3c"

ax_b.plot(pump_ratios, mean_n_curve, color=color_n, linewidth=2.4, label=r"Mean Photons $\langle n \rangle$")
ax_b.axvline(1.0, color="gray", linestyle="--", linewidth=1.8, label=r"Laser Threshold $A = \kappa$")
ax_b.set_xlabel(r"Pump Parameter $A / \kappa$")
ax_b.set_ylabel(r"Mean Photon Number $\langle n \rangle$", color=color_n)
ax_b.tick_params(axis='y', labelcolor=color_n)

# Twin axis for g^(2)(0)
ax_b_twin = ax_b.twinx()
ax_b_twin.plot(pump_ratios, g2_curve, color=color_g2, linewidth=2.4, linestyle="-.", label=r"Coherence $g^{(2)}(0)$")
ax_b_twin.set_ylabel(r"Second-Order Coherence $g^{(2)}(0)$", color=color_g2)
ax_b_twin.tick_params(axis='y', labelcolor=color_g2)
ax_b_twin.set_ylim(0.8, 2.2)

ax_b.set_title(r"(b) Laser Phase Transition: Intensity & Second-Order Coherence", fontweight="bold")

# Combine legends
lines_1, labels_1 = ax_b.get_legend_handles_labels()
lines_2, labels_2 = ax_b_twin.get_legend_handles_labels()
ax_b.legend(lines_1 + lines_2, labels_1 + labels_2, loc="center left", framealpha=0.9)

# -------------------------------------------------------------
# Panel (c): Micromaser Trapping State Truncation
# -------------------------------------------------------------
ax_c = axes[1, 0]
n_micro_axis = np.arange(40)
ax_c.bar(n_micro_axis - 0.2, p_notrap[:40], width=0.4, color="#95a5a6", alpha=0.7, label=r"Untrapped Regular Regime")
ax_c.bar(n_micro_axis + 0.2, p_trap[:40], width=0.4, color="#8e44ad", alpha=0.9, label=rf"Trapping State ($n_q = {nq_target}$)")

ax_c.axvline(nq_target + 0.5, color="#e74c3c", linestyle="--", linewidth=2.0)
ax_c.annotate(r"Trapping Cutoff: $P(n > n_q) \equiv 0$" + "\n" + r"$g\tau_{\rm int}\sqrt{n_q+1} = \pi$", 
              xy=(nq_target + 0.5, 0.08), xytext=(nq_target + 4, 0.12),
              arrowprops=dict(arrowstyle="->", color="#e74c3c", lw=1.5),
              bbox=dict(boxstyle="round,pad=0.4", fc="#f4ecf7", ec="#8e44ad", lw=1.2),
              fontweight="bold")

ax_c.set_title(r"(c) Micromaser Photon Distribution & Trapping Truncation", fontweight="bold")
ax_c.set_xlabel(r"Photon Number $n$")
ax_c.set_ylabel(r"Probability $P(n)$")
ax_c.set_xlim(-1, 35)
ax_c.legend(loc="upper right", framealpha=0.9)

# -------------------------------------------------------------
# Panel (d): Non-Classical Sub-Poissonian Statistics (Mandel Q)
# -------------------------------------------------------------
ax_d = axes[1, 1]
ax_d.plot(theta_vals, Q_micro_curve, color="#16a085", linewidth=2.2, label=r"Micromaser Mandel $Q(\theta)$")
ax_d.axhline(0.0, color="black", linestyle="--", linewidth=1.5, label=r"Poissonian Limit $Q = 0$")
ax_d.fill_between(theta_vals, Q_micro_curve, 0.0, where=(np.array(Q_micro_curve) < 0), 
                  color="#1abc9c", alpha=0.35, label=r"Sub-Poissonian Light ($Q < 0$)")

ax_d.set_title(r"(d) Micromaser Mandel $Q$ vs. Pump Parameter $\theta$", fontweight="bold")
ax_d.set_xlabel(r"Pump Parameter $\theta = g\tau_{\rm int}\sqrt{N_{\rm ex}}$")
ax_d.set_ylabel(r"Mandel Parameter $Q$")
ax_d.set_ylim(-1.1, 4.0)
ax_d.legend(loc="upper right", framealpha=0.9)

plt.tight_layout()
plt.show()

```

> **Physical Insight: The Laser Threshold as a Non-Equilibrium Phase Transition**
>
> 1. **Broken Gauge Symmetry and Coherence:** In Landau phase transition theory, a ferromagnet develops a macroscopic magnetization $\langle M\rangle \ne 0$ below the Curie temperature, spontaneously breaking continuous rotational symmetry. Similarly, above the laser threshold ($A > \kappa$), the cavity electric field spontaneously breaks continuous phase symmetry ($U(1)$ gauge symmetry), acquiring a macroscopic coherent amplitude with fixed phase. Below threshold, the state is an incoherent thermal mixture ($g^{(2)}(0) = 2$); above threshold, it transforms into an intensely coherent state ($g^{(2)}(0) = 1$).
> 2. **Why Do Trapping States Occur Only in the Micromaser?** In standard lasers, macroscopic atom numbers wash out discrete quantum interference effects. In the micromaser, however, each atom interacts with the cavity independently. When an atom's interaction time $\tau_{\text{int}}$ matches an exact integer number of Rabi cycles ($g\tau_{\text{int}}\sqrt{n_q + 1} = q\pi$), the emission probability becomes identically zero. The state $|n_q\rangle$ acts as an impenetrable reflective barrier in Fock space!
> 3. **Sub-Poissonian Non-Classical Light ($Q < 0$):** Classical electromagnetic fields are mathematically constrained to satisfy Mandel parameter $Q \ge 0$ (and $g^{(2)}(0) \ge 1$) by the Cauchy-Schwarz inequality. The observation of negative Mandel parameters $Q < 0$ in the micromaser provides definitive experimental proof of the **quantum nature of light**, demonstrating photon number fluctuations suppressed below the shot-noise limit.
