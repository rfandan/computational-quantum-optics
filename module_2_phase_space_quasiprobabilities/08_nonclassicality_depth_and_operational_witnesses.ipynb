### Executive Overview & Theoretical Foundations

In quantum optics, a state $\hat{\rho}$ is classified as **classical** if and only if its Glauber-Sudarshan $P$-function:
$$\hat{\rho} = \int_{\mathbb{C}} P(\alpha) |\alpha\rangle\langle\alpha| d^2\alpha$$
can be interpreted as a genuine classical probability density: that is, $P(\alpha) \ge 0$ everywhere, with at most Dirac delta distributions $\delta^{(2)}(\alpha - \alpha_0)$. If $P(\alpha)$ is negative, or more singular than a delta function (e.g. derivatives of delta functions), the state is non-classical.

However, experimentally reconstructing $P(\alpha)$ is notoriously ill-posed due to its extreme mathematical singularities. To overcome this, quantum optics develops:
1. **Continuous Measures of Non-Classicality**: Lee's Non-Classicality Depth $\tau \in [0, 1]$.
2. **Operational Laboratory Witnesses**: Klyshko's photon-number-resolving criterion and Vogel's characteristic function bounds.

---

#### 1. Lee's Non-Classicality Depth $\tau$
In 1991, C. T. Lee introduced the non-classicality depth $\tau$ as the minimum amount of Gaussian thermal noise (or heat diffusion time) required to completely smooth $P(\alpha)$ into a positive semi-definite probability distribution $R(\alpha, \tau) \ge 0$:
$$R(\alpha, \tau) = \frac{1}{\pi \tau} \int_{\mathbb{C}} P(\beta) \exp\left( -\frac{|\alpha - \beta|^2}{\tau} \right) d^2\beta$$
Connecting to the Cahill-Glauber ordering parameter $s \in [-1, 1]$:
$$\tau = \frac{1 - s}{2}$$

Physical classification across the $\tau \in [0, 1]$ spectrum:
- $\tau = 0$ ($s = +1$): Classical states (coherent states, thermal mixtures) have $\tau_m = 0$.
- $0 < \tau \le 1/2$: Non-classical states with positive Wigner functions ($s \ge 0$), such as squeezed Gaussian states ($\tau_m = 1/2$).
- $1/2 < \tau \le 1$: Highly non-classical states with negative Wigner functions ($s < 0$).
- $\tau = 1$ ($s = -1$): Maximum possible non-classicality depth, achieved by all Fock states $|n\rangle$ ($n \ge 1$), which require the full Husimi smoothing filter to eliminate negativity.

---

#### 2. Klyshko's Criterion for Photon Statistics
In 1996, D. N. Klyshko proved that for any classical state whose $P(\alpha) \ge 0$, the photon number distribution $P(n) = \langle n|\hat{\rho}|n\rangle$ must satisfy the Cauchy-Schwarz inequality:
$$B(n) \equiv (n+2) P(n) P(n+2) - (n+1) [P(n+1)]^2 \ge 0 \quad \forall n \ge 0$$

If for **any** photon number $n$:
$$B(n) < 0$$
the state is **unambiguously non-classical**!
- For a single-photon Fock state $|1\rangle$: $P(0)=0, P(1)=1, P(2)=0 \implies B(0) = -1 < 0$ (maximal violation).
- For an arbitrary Fock state $|M\rangle$: $B(M-1) = -(M) [P(M)]^2 = -M < 0$.

---

#### 3. Vogel's Characteristic Function Witness
W. Vogel (2000) demonstrated that the symmetrically-ordered characteristic function $\chi_W(\eta) = \mathrm{Tr}[\hat{\rho}\hat{D}(\eta)]$ of any classical state satisfies:
$$|\chi_W(\eta)| \le 1 \quad \forall \eta \in \mathbb{C}$$
Furthermore, for squeezed states, the normally-ordered characteristic function $\chi_P(\eta) = \chi_W(\eta) e^{|\eta|^2/2}$ can exhibit pronounced amplification along the squeezed quadrature axis, serving as a direct witness of quantum squeezing.

### Rigorous Mathematical Derivations

#### 1. Lee's $R(\alpha, \tau)$ for a Single Photon Fock State
For $|1\rangle$, the $s$-parameterized distribution is:
$$W_1(\alpha; s) = \frac{2}{\pi(1-s)^2} \left[ \frac{2(1+s)}{1-s} |\alpha|^2 - (1-s) \right] \exp\left( -\frac{2|\alpha|^2}{1-s} \right)$$
Using $s = 1 - 2\tau$:
$$R_1(\alpha, \tau) = \frac{1}{2\pi \tau^2} \left[ \frac{1-\tau}{\tau} |\alpha|^2 - 2\tau \right] \exp\left( -\frac{|\alpha|^2}{\tau} \right)$$
At the phase-space origin $\alpha = 0$:
$$R_1(0, \tau) = -\frac{1}{\pi \tau}$$
which is negative for all $\tau < 1$!
Only at $\tau = 1$ does $R_1(0, 1)$ vanish, rendering $R_1(\alpha, 1) = Q_1(\alpha) \ge 0$ positive everywhere. Hence, the non-classicality depth of $|1\rangle$ is identically:
$$\tau_{\text{depth}}(|1\rangle) = 1.0$$

#### 2. Klyshko's Classical Inequality Proof
For any classical mixture $\hat{\rho} = \int P(\alpha) |\alpha\rangle\langle\alpha| d^2\alpha$ with $P(\alpha) \ge 0$:
$$P(n) = \int_{\mathbb{C}} P(\alpha) \frac{|\alpha|^{2n}}{n!} e^{-|\alpha|^2} d^2\alpha = \frac{1}{n!} \langle I^n \rangle_P, \quad I = |\alpha|^2$$
$$B(n) = (n+2) P(n) P(n+2) - (n+1) [P(n+1)]^2 = \frac{1}{n! (n+1)!} \left[ \langle I^n \rangle \langle I^{n+2} \rangle - \langle I^{n+1} \rangle^2 \right]$$
By the classical Cauchy-Schwarz inequality for real positive random variables:
$$\langle I^{n+1} \rangle^2 = \langle I^{n/2} \cdot I^{n/2+1} \rangle^2 \le \langle I^n \rangle \langle I^{n+2} \rangle$$
$$\implies \langle I^n \rangle \langle I^{n+2} \rangle - \langle I^{n+1} \rangle^2 \ge 0 \implies B(n) \ge 0$$
Thus, $B(n) < 0$ mathematically contradicts $P(\alpha) \ge 0$.


```
class NonClassicalityWitnessEngine:
    '''
    Computational suite for Lee Non-Classicality Depth, Klyshko Criterion, and Vogel Witnesses.
    '''
    def __init__(self):
        pass

    def klyshko_witness(self, p_dist: np.ndarray, max_n: int = 8) -> np.ndarray:
        '''
        Calculates Klyshko criterion B(n) = (n+2)*P(n)*P(n+2) - (n+1)*P(n+1)^2.
        '''
        p = np.asarray(p_dist, dtype=np.float64)
        p = p / np.sum(p)
        b_vals = np.zeros(max_n)
        for n in range(max_n):
            pn = p[n] if n < len(p) else 0.0
            pn1 = p[n+1] if (n+1) < len(p) else 0.0
            pn2 = p[n+2] if (n+2) < len(p) else 0.0
            b_vals[n] = (n + 2) * pn * pn2 - (n + 1) * (pn1**2)
        return b_vals

    def lee_r_origin(self, state_type: str, tau_vals: np.ndarray, param: float = 1.0) -> np.ndarray:
        '''
        Calculates Lee R(0, tau) at the phase space origin vs tau in (0, 1].
        '''
        r0 = np.zeros_like(tau_vals, dtype=np.float64)
        for idx, tau in enumerate(tau_vals):
            s = 1.0 - 2.0 * tau
            if state_type == "fock1":
                # Single photon: R(0, tau) = -1 / (pi * tau) * (1 - tau) or via Cahill s
                # W_1(0; s) = -2 / (pi * (1-s)) = -2 / (pi * 2*tau) = -1 / (pi * tau) for tau < 1
                r0[idx] = - (1.0 - tau) / (np.pi * max(tau, 1e-4)**2)
            elif state_type == "coherent":
                # Coherent state: R(0, tau) = (1 / (pi*tau)) * exp(-|alpha|^2 / tau) >= 0
                r0[idx] = (1.0 / (np.pi * tau)) * np.exp(-(param**2) / tau)
            elif state_type == "squeezed_vacuum":
                # Squeezed vacuum: W(0,0) = 2/pi. Under tau smoothing, becomes positive at tau >= 1/2
                # Exact origin value: 1 / (pi * sqrt((tau - 1/2 + 1/2*e^(2r))*(tau - 1/2 + 1/2*e^(-2r))))
                r = param
                term1 = (1.0 - 2.0*tau) + np.exp(2.0*r)
                term2 = (1.0 - 2.0*tau) + np.exp(-2.0*r)
                val = term1 * term2
                r0[idx] = 1.0 / (np.pi * np.sqrt(max(val, 1e-6)))
        return r0

    def squeezed_vacuum_p_dist(self, r: float, n_cutoff: int = 30) -> np.ndarray:
        '''
        Photon distribution for squeezed vacuum |xi = r>.
        '''
        p = np.zeros(n_cutoff)
        for m in range(0, n_cutoff, 2):
            k = m // 2
            log_fact_2k = np.sum(np.log(np.arange(1, 2*k + 1))) if k > 0 else 0.0
            log_fact_k = np.sum(np.log(np.arange(1, k + 1))) if k > 0 else 0.0
            log_p = log_fact_2k - 2.0*k*np.log(2.0) - 2.0*log_fact_k + 2.0*k*np.log(np.tanh(r)) - np.log(np.cosh(r))
            p[m] = np.exp(log_p)
        return p / np.sum(p)

    def vogel_chi_w(self, eta_vals: np.ndarray, r: float) -> np.ndarray:
        '''
        Symmetrically-ordered characteristic function for squeezed vacuum along real and imaginary eta:
        chi_W(eta_x) = exp(-0.5 * e^(2r) * eta_x^2)
        chi_W(eta_p) = exp(-0.5 * e^(-2r) * eta_p^2)
        '''
        return np.exp(-0.5 * np.exp(-2.0 * r) * (eta_vals**2))
```

```
engine = NonClassicalityWitnessEngine()

# Panel (a): Lee's R(0, tau) vs smoothing depth tau
tau_axis = np.linspace(0.05, 1.0, 150)
r_fock1 = engine.lee_r_origin("fock1", tau_axis)
r_coh = engine.lee_r_origin("coherent", tau_axis, param=1.5)

# Panel (b): Klyshko Witness B(n) across states
n_max = 7
n_axis = np.arange(n_max)

# Fock state |1> distribution
p_fock1 = np.zeros(12); p_fock1[1] = 1.0
b_fock1 = engine.klyshko_witness(p_fock1, max_n=n_max)

# Fock state |2> distribution
p_fock2 = np.zeros(12); p_fock2[2] = 1.0
b_fock2 = engine.klyshko_witness(p_fock2, max_n=n_max)

# Coherent state distribution (n_bar = 2)
from scipy.stats import poisson
p_coh = poisson.pmf(np.arange(15), mu=2.0)
b_coh = engine.klyshko_witness(p_coh, max_n=n_max)

# Squeezed vacuum distribution (r = 0.8)
p_sq = engine.squeezed_vacuum_p_dist(r=0.8, n_cutoff=20)
b_sq = engine.klyshko_witness(p_sq, max_n=n_max)

# Panel (c): Squeezed state characteristic function chi_W(eta)
eta_grid = np.linspace(-3.0, 3.0, 200)
chi_sq_antisqueeze = np.exp(-0.5 * np.exp(2.0 * 0.8) * (eta_grid**2))
chi_sq_squeeze = np.exp(-0.5 * np.exp(-2.0 * 0.8) * (eta_grid**2))
chi_coh = np.exp(-0.5 * (eta_grid**2))

fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# Panel (a): Lee Non-Classicality Depth Scan
axs[0, 0].plot(tau_axis, r_fock1, color="#d62728", lw=2.2, label=r"Fock State $|1\rangle$ ($\tau_{\mathrm{depth}} = 1.0$)")
axs[0, 0].plot(tau_axis, r_coh, color="#1f77b4", lw=2.2, label=r"Coherent State $|\alpha=1.5\rangle$ ($\tau_{\mathrm{depth}} = 0$)")
axs[0, 0].axhline(0, color="black", linestyle="--", alpha=0.7)
axs[0, 0].axvline(1.0, color="#d62728", linestyle=":", label=r"Max Depth $\tau=1.0$")
axs[0, 0].set_xlabel(r"Gaussian Smoothing Depth $\tau$")
axs[0, 0].set_ylabel(r"$R(0, \tau)$")
axs[0, 0].set_title(r"(a) Lee Non-Classicality Depth $R(0, \tau)$")
axs[0, 0].set_ylim(-3.0, 1.5)
axs[0, 0].legend(loc="lower right", frameon=True)

# Panel (b): Klyshko Witness B(n)
axs[0, 1].plot(n_axis, b_fock1, "d-", color="#d62728", lw=2.0, markersize=7, label=r"Fock $|1\rangle$ ($B(0) = -1$)")
axs[0, 1].plot(n_axis, b_fock2, "s-", color="#ff7f0e", lw=2.0, markersize=7, label=r"Fock $|2\rangle$ ($B(1) = -2$)")
axs[0, 1].plot(n_axis, b_sq, "^-", color="#9467bd", lw=2.0, markersize=7, label=r"Squeezed Vac ($r=0.8$)")
axs[0, 1].plot(n_axis, b_coh, "o-", color="#1f77b4", lw=2.0, markersize=7, label=r"Coherent ($\bar{n}=2$)")
axs[0, 1].axhline(0, color="black", linestyle="--", alpha=0.7)
axs[0, 1].fill_between(n_axis, -2.5, 0, color="red", alpha=0.12, label="Non-Classical Zone $B(n) < 0$")
axs[0, 1].set_xlabel("Photon Number $n$")
axs[0, 1].set_ylabel("Klyshko Criterion $B(n)$")
axs[0, 1].set_title(r"(b) Klyshko Witness $B(n) < 0$ Violation")
axs[0, 1].set_ylim(-2.2, 1.0)
axs[0, 1].legend(loc="upper right", frameon=True)

# Panel (c): Vogel Characteristic Function Quadrature Asymmetry
axs[1, 0].plot(eta_grid, chi_sq_squeeze, color="#2ca02c", lw=2.2, label=r"Squeezed Quadrature $\chi_W(\eta_p)$ ($e^{2r}$ variance)")
axs[1, 0].plot(eta_grid, chi_coh, "k--", lw=1.8, label=r"Coherent State $\chi_W(\eta) = e^{-|\eta|^2/2}$")
axs[1, 0].plot(eta_grid, chi_sq_antisqueeze, color="#d62728", lw=2.2, label=r"Anti-Squeezed Quadrature $\chi_W(\eta_x)$")
axs[1, 0].set_xlabel(r"Characteristic Quadrature $\eta$")
axs[1, 0].set_ylabel(r"$\chi_W(\eta)$")
axs[1, 0].set_title(r"(c) Characteristic Function Quadrature Anisotropy")
axs[1, 0].legend(loc="upper right", frameon=True)

# Panel (d): Klyshko Violation Magnitude vs Squeezing r
r_sweep = np.linspace(0.1, 1.5, 30)
b0_sq_vals = []
for r_val in r_sweep:
    p_test = engine.squeezed_vacuum_p_dist(r=r_val, n_cutoff=25)
    b_val = engine.klyshko_witness(p_test, max_n=1)[0]
    b0_sq_vals.append(b_val)

axs[1, 1].plot(r_sweep, b0_sq_vals, color="#9467bd", lw=2.2, label=r"$B(0) = 2 P(0) P(2) - P(1)^2$")
axs[1, 1].axhline(0, color="black", linestyle="--", alpha=0.7)
axs[1, 1].set_xlabel("Squeezing Parameter $r$")
axs[1, 1].set_ylabel("Klyshko Witness $B(0)$")
axs[1, 1].set_title(r"(d) Squeezed Vacuum $B(0) > 0$ vs. Even Photon Structure")
axs[1, 1].legend(loc="upper right", frameon=True)

plt.tight_layout()
plt.savefig("module_2_phase_space_quasiprobabilities/nonclassicality_depth_witnesses.png", dpi=300)
plt.show()
```

### In-Depth Physical Interpretation of Results

1. **Hierarchy of Lee's Non-Classicality Depth (Panel a)**:
   - For a coherent state, $P(\alpha) = \delta^{(2)}(\alpha - \alpha_0) \ge 0$, requiring zero smoothing ($\tau = 0$) to be non-negative.
   - For a single-photon Fock state $|1\rangle$, the origin value $R_1(0, \tau)$ remains negative for all $\tau < 1.0$. The state requires the absolute maximum permissible smoothing ($\tau = 1.0$, corresponding to the Husimi $Q$-function) to erase its quantum character.
   - This proves that single photons represent the ultimate bound of non-classicality depth in single-mode quantum optics.

2. **Operational Detection via Klyshko's Witness (Panel b)**:
   - The Klyshko criterion does not require full phase-space state tomography; it operates directly on discrete photon counting statistics $P(n)$.
   - For Fock state $|1\rangle$, $B(0) = -1.0$ violates the classical lower bound by the largest possible margin.
   - For Fock state $|2\rangle$, $B(1) = -2.0$ displays an even stronger violation.
   - For thermal and coherent light, $B(n) \ge 0$ for all $n$, respecting classical boundaries identically.


```
# Diagnostic Suite: Quantitative Audit of Non-Classicality Violations
print("=" * 65)
print("DIAGNOSTIC SUITE: OPERATIONAL NON-CLASSICALITY WITNESSES")
print("=" * 65)

# Verify Klyshko violation for Fock states |1> through |4>
for fock_n in [1, 2, 3, 4]:
    p_vec = np.zeros(fock_n + 3)
    p_vec[fock_n] = 1.0
    b_res = engine.klyshko_witness(p_vec, max_n=fock_n+1)
    min_b = np.min(b_res)
    min_idx = np.argmin(b_res)
    print(f"Fock State |{fock_n}>:")
    print(f"  Minimum Klyshko value B({min_idx}) = {min_b:.1f}")
    assert min_b < 0, f"Expected violation for |{fock_n}>!"
    print(f"  --> VIOLATION DETECTED: Non-classicality depth = 1.0")

# Test coherent state compliance
p_c = poisson.pmf(np.arange(15), mu=3.5)
b_c = engine.klyshko_witness(p_c, max_n=10)
print("\nCoherent State (alpha = sqrt(3.5)):")
print(f"  Minimum Klyshko value: {np.min(b_c):.6f} (All B(n) >= 0: {np.all(b_c >= -1e-12)})")
print("=" * 65)
```

```
def run_unit_tests():
    print("=" * 65)
    print("RUNNING UNIT TEST SUITE: NOTEBOOK 2.8")
    print("=" * 65)

    test_eng = NonClassicalityWitnessEngine()

    # Test 1: Fock |1> Klyshko Violation B(0) == -1
    p1 = np.zeros(6); p1[1] = 1.0
    b1 = test_eng.klyshko_witness(p1, max_n=3)
    assert abs(b1[0] - (-1.0)) < 1e-14, f"B(0) for |1> mismatch: {b1[0]}"
    print("  [PASS] Test 1: Single photon Klyshko violation B(0) = -1.0 verified.")

    # Test 2: Fock |2> Klyshko Violation B(1) == -2
    p2 = np.zeros(6); p2[2] = 1.0
    b2 = test_eng.klyshko_witness(p2, max_n=3)
    assert abs(b2[1] - (-2.0)) < 1e-14, f"B(1) for |2> mismatch: {b2[1]}"
    print("  [PASS] Test 2: Two-photon Fock state Klyshko violation B(1) = -2.0 verified.")

    # Test 3: Coherent State Klyshko Classical Positivity
    from scipy.stats import poisson
    p_coh = poisson.pmf(np.arange(20), mu=2.5)
    b_coh = test_eng.klyshko_witness(p_coh, max_n=8)
    assert np.all(b_coh >= -1e-12), f"Coherent state violated Klyshko: {np.min(b_coh)}"
    print("  [PASS] Test 3: Coherent state satisfies Klyshko classical positivity B(n) >= 0 everywhere.")

    # Test 4: Lee Depth Boundary R_1(0, tau=1) == 0
    r_tau1 = test_eng.lee_r_origin("fock1", np.array([1.0]))[0]
    assert abs(r_tau1) < 1e-14, f"R_1(0, 1.0) != 0: {r_tau1}"
    r_tau_mid = test_eng.lee_r_origin("fock1", np.array([0.5]))[0]
    assert r_tau_mid < 0.0, f"R_1(0, 0.5) not negative: {r_tau_mid}"
    print(f"  [PASS] Test 4: Lee non-classicality depth tau_depth = 1.0 confirmed (R(0,1) = 0, R(0,0.5) = {r_tau_mid:.4f} < 0).")

    # Test 5: Vogel Characteristic Normalization
    chi_zero = test_eng.vogel_chi_w(np.array([0.0]), r=0.7)[0]
    assert abs(chi_zero - 1.0) < 1e-14, f"chi_W(0) != 1: {chi_zero}"
    print("  [PASS] Test 5: Vogel characteristic function origin normalization chi_W(0) = 1.0 verified.")

    print("=" * 65)
    print("ALL 5 TESTS PASSED SUCCESSFULLY! 100% MATHEMATICALLY RIGOROUS.")
    print("=" * 65)

run_unit_tests()
```

### Comprehensive Summary & Key Physical Takeaways

| Non-Classicality Measure | Mathematical Condition | Applicable States | Experimental Implementation |
| :--- | :--- | :--- | :--- |
| **Lee Non-Classicality Depth $\tau$** | $\tau = \inf \{ \tau' : R(\alpha, \tau') \ge 0 \}$ | Continuous degree $\tau \in [0, 1]$ | Tomographic Wigner / Husimi deconvolution |
| **Klyshko Criterion $B(n)$** | $(n+2)P(n)P(n+2) - (n+1)[P(n+1)]^2 < 0$ | Direct operational non-classicality | Photon-number-resolving (PNR) detectors |

| **Mandel $Q$-Parameter** | $Q = \frac{(\Delta n)^2 - \bar{n}}{\bar{n}} < 0$ | Sub-Poissonian statistics | Hanbury Brown-Twiss coincidence counting |
