## Theoretical Framework: Two-Mode Squeezing & The CV EPR Paradox

### 1. The Two-Mode Squeezing Operator
In non-degenerate optical parametric amplification (NOPA) or spontaneous parametric down-conversion (SPDC), a high-energy pump photon splits into a signal photon $\hat{a}$ and an idler photon $\hat{b}$. 
The interaction Hamiltonian is given by:
$$\hat{H}_{\text{int}} = i\hbar \chi \left( \hat{a}^\dagger \hat{b}^\dagger - \hat{a}\hat{b} \right)$$
which generates the **two-mode squeezing operator**:
$$\hat{S}_2(r) = \exp\left[ r (\hat{a}\hat{b} - \hat{a}^\dagger \hat{b}^\dagger) \right]$$
where $r = \chi t$ is the dimensionless squeezing parameter.

The operator transforms mode annihilation operators via the Bogoliubov transformation:
$$\hat{S}_2^\dagger(r) \hat{a} \hat{S}_2(r) = \hat{a}\cosh r - \hat{b}^\dagger\sinh r$$
$$\hat{S}_2^\dagger(r) \hat{b} \hat{S}_2(r) = \hat{b}\cosh r - \hat{a}^\dagger\sinh r$$

---

### 2. The Two-Mode Squeezed Vacuum (TMSV)
Acting with $\hat{S}_2(r)$ on the joint vacuum state $|0, 0\rangle$ yields the **Two-Mode Squeezed Vacuum (TMSV)** state (also known as the twin-beam state):
$$|\text{TMSV}(r)\rangle = \hat{S}_2(r)|0, 0\rangle = \frac{1}{\cosh r}\sum_{n=0}^{\infty} (\tanh r)^n |n, n\rangle$$

#### Key Properties of TMSV:
1. **Twin-Photon Number Correlations:** Photons are generated strictly in pairs! The probability of finding $n_a$ photons in mode $a$ and $n_b$ photons in mode $b$ is strictly diagonal:
   $$P(n_a, n_b) = \frac{(\tanh r)^{2n_a}}{\cosh^2 r} \delta_{n_a, n_b}$$
2. **Mean Photon Number per Mode:**
   $$\bar{n}_a = \langle \hat{a}^\dagger \hat{a} \rangle = \sinh^2 r, \quad \bar{n}_b = \langle \hat{b}^\dagger \hat{b} \rangle = \sinh^2 r$$
3. **Total Photon Number:**
   $$\langle \hat{N}_{\text{tot}} \rangle = 2\sinh^2 r$$

---

### 3. The Continuous-Variable Einstein-Podolsky-Rosen (EPR) Paradox
In their historic 1935 paper, Einstein, Podolsky, and Rosen considered two particles with perfectly correlated relative position and total momentum:
$$\hat{x}_- = \frac{\hat{x}_a - \hat{x}_b}{\sqrt{2}}, \quad \hat{p}_+ = \frac{\hat{p}_a + \hat{p}_b}{\sqrt{2}}$$
where $\hat{x}_k = (\hat{a}_k + \hat{a}_k^\dagger)/\sqrt{2}$ and $\hat{p}_k = (\hat{a}_k - \hat{a}_k^\dagger)/(i\sqrt{2})$ with $[\hat{x}_k, \hat{p}_k] = i$.

Crucially, these composite operators commute:
$$[\hat{x}_-, \hat{p}_+] = \frac{1}{2}\left( [\hat{x}_a, \hat{p}_a] - [\hat{x}_b, \hat{p}_b] \right) = \frac{1}{2}(i - i) \equiv 0!$$
They can be measured simultaneously without any Heisenberg uncertainty limit between them!

In the TMSV state, calculating the quantum variances yields:
$$\Delta x_-^2 = \langle \hat{x}_-^2 \rangle - \langle \hat{x}_- \rangle^2 = \frac{1}{2} e^{-2r}$$
$$\Delta p_+^2 = \langle \hat{p}_+^2 \rangle - \langle \hat{p}_+ \rangle^2 = \frac{1}{2} e^{-2r}$$

In the limit $r \to \infty$, both variances vanish exponentially:
$$\lim_{r \to \infty} \Delta x_-^2 = 0, \quad \lim_{r \to \infty} \Delta p_+^2 = 0$$
Measuring $\hat{x}_a$ predicts $\hat{x}_b$ with absolute certainty, while measuring $\hat{p}_a$ predicts $\hat{p}_b$ with absolute certainty, realizing the exact physical embodiment of the EPR state.

---

### 4. The Duan-Giedke-Cirac-Zoller (DGCZ) Inseparability Criterion
For any separable quantum state $\hat{\rho}_{\text{sep}} = \sum_i p_i \hat{\rho}_a^{(i)} \otimes \hat{\rho}_b^{(i)}$, the sum of relative variances is bounded by the standard quantum limit (Duan et al., Simon 2000):
$$\Delta x_-^2 + \Delta p_+^2 \ge 1.0 \quad (\text{Separability Bound})$$

For the Two-Mode Squeezed Vacuum:
$$\Delta x_-^2 + \Delta p_+^2 = \frac{1}{2}e^{-2r} + \frac{1}{2}e^{-2r} = e^{-2r}$$
For any non-zero squeezing parameter $r > 0$:
$$e^{-2r} < 1.0 \implies \text{Entangled!}$$
Any observed variance sum below $1.0$ is an unequivocal, sufficient witness of continuous-variable bipartite quantum entanglement.

---

### 5. Thermal Subsystems & Entanglement Entropy
Tracing out mode $b$ reveals the reduced state of mode $a$:
$$\hat{\rho}_a = \text{Tr}_b(|\text{TMSV}\rangle\langle\text{TMSV}|) = \frac{1}{\cosh^2 r}\sum_{n=0}^{\infty} (\tanh^2 r)^n |n\rangle\langle n|$$

This is an **exact single-mode thermal state** with mean photon number $\bar{n} = \sinh^2 r$!
The bipartite von Neumann entanglement entropy is given by:
$$S(\hat{\rho}_a) = (\bar{n} + 1)\ln(\bar{n} + 1) - \bar{n}\ln\bar{n} = \cosh^2 r \ln(\cosh^2 r) - \sinh^2 r \ln(\sinh^2 r)$$
As $r$ increases, the entanglement entropy grows monotonically and asymptotically as $S(r) \sim 2r$.



```python
# Implementation: High-Precision Two-Mode Squeezing & EPR Engine

class TwoModeSqueezingEngine:
    """
    Simulation engine for Two-Mode Squeezed Vacuum (TMSV) and continuous-variable EPR states.
    Supports exact state construction in the Fock basis, quadrature variance calculations,
    and DGCZ inseparability verification.
    """
    def __init__(self, n_cutoff=25):
        self.cutoff = int(n_cutoff)
        self.dim = self.cutoff * self.cutoff
        
        # Single-mode operators
        a_single = np.zeros((self.cutoff, self.cutoff), dtype=complex)
        for n in range(1, self.cutoff):
            a_single[n-1, n] = np.sqrt(n)
        Id_single = np.eye(self.cutoff, dtype=complex)
        
        # Two-mode operators
        self.a = np.kron(a_single, Id_single)
        self.b = np.kron(Id_single, a_single)
        self.a_dag = self.a.conj().T
        self.b_dag = self.b.conj().T
        
        # Quadratures: x_k = (a_k + a_k^dag)/sqrt(2), p_k = (a_k - a_k^dag)/(i*sqrt(2))
        sqrt2 = np.sqrt(2.0)
        self.xa = (self.a + self.a_dag) / sqrt2
        self.pa = (self.a - self.a_dag) / (1j * sqrt2)
        self.xb = (self.b + self.b_dag) / sqrt2
        self.pb = (self.b - self.b_dag) / (1j * sqrt2)
        
        # Relative EPR quadratures
        self.x_minus = (self.xa - self.xb) / sqrt2
        self.p_plus = (self.pa + self.pb) / sqrt2
        
        # Precompute operator squares for high-speed variance computation
        self.x_minus_sq = np.dot(self.x_minus, self.x_minus)
        self.p_plus_sq = np.dot(self.p_plus, self.p_plus)

    def state_tmsv(self, r):
        """
        Construct the Two-Mode Squeezed Vacuum state vector in Fock basis:
            |TMSV(r)> = (1 / cosh(r)) * sum_n (tanh(r))^n |n, n>
        """
        r_val = float(r)
        sech_r = 1.0 / np.cosh(r_val)
        tanh_r = np.tanh(r_val)
        
        psi = np.zeros(self.dim, dtype=complex)
        for n in range(self.cutoff):
            idx = n * self.cutoff + n
            psi[idx] = sech_r * (tanh_r**n)
            
        # Re-normalize to compensate for finite Fock truncation
        psi /= la.norm(psi)
        return psi

    def joint_photon_distribution(self, psi):
        """Compute the 2D joint photon number probability distribution P(n_a, n_b)."""
        probs = np.abs(psi)**2
        return probs.reshape((self.cutoff, self.cutoff))

    def dgcz_variance_sum(self, state):
        """Compute the DGCZ inseparability variance sum: Delta x_-^2 + Delta p_+^2."""
        # Using precomputed operator squares for maximum vector speed
        exp_xm = float(np.real(np.dot(state.conj().T, np.dot(self.x_minus, state))))
        exp_xm2 = float(np.real(np.dot(state.conj().T, np.dot(self.x_minus_sq, state))))
        var_xm = exp_xm2 - exp_xm**2
        
        exp_pp = float(np.real(np.dot(state.conj().T, np.dot(self.p_plus, state))))
        exp_pp2 = float(np.real(np.dot(state.conj().T, np.dot(self.p_plus_sq, state))))
        var_pp = exp_pp2 - exp_pp**2
        
        return var_xm + var_pp, var_xm, var_pp

    def entanglement_entropy(self, psi):
        """Compute von Neumann entanglement entropy S(rho_a) from Schmidt coefficients."""
        psi_mat = psi.reshape((self.cutoff, self.cutoff))
        rho_a = np.dot(psi_mat, psi_mat.conj().T)
        evals = np.real(la.eigvalsh(rho_a))
        pos_evals = evals[evals > 1e-15]
        return -float(np.sum(pos_evals * np.log(pos_evals)))

print("TwoModeSqueezingEngine successfully loaded.")

```

```python
# Publication-Quality 4-Panel Visualization: TMSV Photon Correlations & EPR Inseparability

engine = TwoModeSqueezingEngine(n_cutoff=25)

# 1. State at squeezing r = 0.8
r_example = 0.8
psi_tmsv_ex = engine.state_tmsv(r_example)
P_joint_ex = engine.joint_photon_distribution(psi_tmsv_ex)

# 2. Squeezing scan for DGCZ inseparability and Entanglement Entropy
r_scan = np.linspace(0.0, 1.8, 60)
dgcz_sum_list = []
var_xm_list = []
var_pp_list = []
entropy_list = []

for r in r_scan:
    psi_r = engine.state_tmsv(r)
    v_sum, vx, vp = engine.dgcz_variance_sum(psi_r)
    dgcz_sum_list.append(v_sum)
    var_xm_list.append(vx)
    var_pp_list.append(vp)
    entropy_list.append(engine.entanglement_entropy(psi_r))

# Analytical curves
dgcz_analytic = np.exp(-2.0 * r_scan)
n_bar_analytic = np.sinh(r_scan)**2
entropy_analytic = np.cosh(r_scan)**2 * np.log(np.cosh(r_scan)**2) - np.sinh(r_scan)**2 * np.log(np.maximum(np.sinh(r_scan)**2, 1e-300))

# 3. Reduced state thermal distribution comparison for r = 0.8
p_marginal_a = np.sum(P_joint_ex, axis=1)
n_axis = np.arange(15)
n_bar_ex = np.sinh(r_example)**2
p_thermal_analytic = (1.0 / (1.0 + n_bar_ex)) * (n_bar_ex / (1.0 + n_bar_ex))**n_axis

# Generate 4-Panel Masterpiece Figure
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# -------------------------------------------------------------
# Panel (a): Joint Photon Number Distribution P(na, nb)
# -------------------------------------------------------------
ax_a = axes[0, 0]
im = ax_a.imshow(P_joint_ex[:12, :12], cmap="viridis", origin="lower", extent=[-0.5, 11.5, -0.5, 11.5])
cbar = plt.colorbar(im, ax=ax_a, fraction=0.046, pad=0.04)
cbar.set_label("Joint Probability $P(n_a, n_b)$")

# Highlight diagonal twin-beam correlation
ax_a.plot([-0.5, 11.5], [-0.5, 11.5], color="cyan", linestyle="--", linewidth=1.8, label=r"Twin Photons $n_a = n_b$")
ax_a.set_title(r"(a) Joint Photon Distribution for TMSV ($r = 0.8$)", fontweight="bold")
ax_a.set_xlabel(r"Mode $b$ Photon Number $n_b$")
ax_a.set_ylabel(r"Mode $a$ Photon Number $n_a$")
ax_a.legend(loc="upper left", framealpha=0.9)

# -------------------------------------------------------------
# Panel (b): DGCZ Inseparability Criterion & EPR Squeezing
# -------------------------------------------------------------
ax_b = axes[0, 1]
ax_b.plot(r_scan, dgcz_sum_list, "o", color="#e74c3c", markersize=5, label=r"Numerical $\Delta x_-^2 + \Delta p_+^2$")
ax_b.plot(r_scan, dgcz_analytic, color="#2c3e50", linewidth=2.2, label=r"Theory $e^{-2r}$")
ax_b.axhline(1.0, color="black", linestyle="--", linewidth=1.8, label=r"Separability Limit ($\geq 1.0$)")

# Shaded entanglement region
ax_b.fill_between(r_scan, 0.0, dgcz_analytic, color="#2ecc71", alpha=0.25, label=r"Entangled Region ($e^{-2r} < 1$)")

ax_b.set_title(r"(b) DGCZ Continuous-Variable Inseparability Witness", fontweight="bold")
ax_b.set_xlabel(r"Squeezing Parameter $r$")
ax_b.set_ylabel(r"Variance Sum $\Delta x_-^2 + \Delta p_+^2$")
ax_b.set_ylim(-0.02, 1.15)
ax_b.legend(loc="upper right", framealpha=0.9)

# -------------------------------------------------------------
# Panel (c): Bipartite Von Neumann Entanglement Entropy
# -------------------------------------------------------------
ax_c = axes[1, 0]
squeezing_dB = 8.686 * r_scan  # 10 * log10(exp(2r))

ax_c.plot(squeezing_dB, entropy_list, "o", color="#8e44ad", markersize=5, label=r"Numerical $S(\hat{\rho}_a)$")
ax_c.plot(squeezing_dB, entropy_analytic, color="#2980b9", linewidth=2.2, label=r"Thermal Entropy $S(r)$")

ax_c.set_title(r"(c) Bipartite Entanglement Entropy vs. Squeezing (dB)", fontweight="bold")
ax_c.set_xlabel(r"Squeezing Level (dB)")
ax_c.set_ylabel(r"Von Neumann Entropy $S(\hat{\rho}_a)$ (nats)")
ax_c.legend(loc="upper left", framealpha=0.9)

# -------------------------------------------------------------
# Panel (d): Reduced State Thermal Statistics
# -------------------------------------------------------------
ax_d = axes[1, 1]
ax_d.bar(n_axis - 0.15, p_marginal_a[:15], width=0.3, color="#e67e22", alpha=0.85, label=r"Reduced Mode $a$ $P(n_a)$")
ax_d.plot(n_axis + 0.15, p_thermal_analytic, "kd", markersize=6, label=r"Bose-Einstein Thermal $(\bar{n} \approx 0.79)$")

ax_d.annotate(r"Exact Thermal Subsystem:" + "\n" + r"$P(n) = \frac{\bar{n}^n}{(1+\bar{n})^{n+1}}$", 
              xy=(2.0, 0.15), xytext=(4.0, 0.28),
              arrowprops=dict(arrowstyle="->", color="black", lw=1.5),
              bbox=dict(boxstyle="round,pad=0.4", fc="#fff9c4", ec="#f39c12", lw=1.2),
              fontweight="bold")

ax_d.set_title(r"(d) Single-Mode Thermalization via Entanglement", fontweight="bold")
ax_d.set_xlabel(r"Measured Photons $n_a$")
ax_d.set_ylabel(r"Probability $P(n_a)$")
ax_d.set_xlim(-0.6, 14.5)
ax_d.legend(loc="upper right", framealpha=0.9)

plt.tight_layout()
plt.show()

```

> **Physical Insight: Entanglement, Thermalization, and the Continuous-Variable EPR State**
>
> 1. **How Can Pure States Yield Thermal Subsystems?** The global two-mode state $|\text{TMSV}\rangle$ is $100\%$ pure ($\text{Tr}(\hat{\rho}_{ab}^2) \equiv 1$). However, because mode $a$ and mode $b$ are intensely entangled, tracing out mode $b$ discards quantum information, leaving mode $a$ in an exact mixed **canonical Gibbs thermal state**:
>    $$\hat{\rho}_a = \frac{1}{Z}\exp\left(-\frac{\hbar\omega \hat{a}^\dagger\hat{a}}{k_B T_{\text{eff}}}\right), \quad k_B T_{\text{eff}} = \frac{\hbar\omega}{\ln(1 + 1/\bar{n})}$$
>    This is a profound manifestation of **canonical thermalization from pure quantum entanglement**: the effective temperature arises solely from missing quantum correlation information!
> 2. **The DGCZ Witness as an Entanglement Gauge:** In classical physics, fluctuating noise in two separate channels is strictly positive and uncorrelated; the relative variances can never sum to less than the shot noise limit $1.0$. The observation that $\Delta x_-^2 + \Delta p_+^2 = e^{-2r} < 1.0$ proves that the two fields fluctuate *in tandem* below the vacuum noise floor, defying classical local realism.
> 3. **The Resource for CV Teleportation:** A two-mode squeezed vacuum is the continuous-variable analogue of an Einstein-Podolsky-Rosen Bell state. In Notebook 5.4, we will harness this exact state to perform **continuous-variable quantum teleportation** of unknown quantum states of light.

```python
# Numerical Verification of Challenges 1, 2, and 3: TMSV & EPR Benchmarks

tmsv_engine = TwoModeSqueezingEngine(n_cutoff=25)

# -------------------------------------------------------------
# Part 1: Twin-Photon Strict Correlation (Challenge 1)
# -------------------------------------------------------------
r_test = 0.85
psi_tmsv_test = tmsv_engine.state_tmsv(r_test)
P_joint_test = tmsv_engine.joint_photon_distribution(psi_tmsv_test)

# Compute off-diagonal probability sum
diag_mask = np.eye(tmsv_engine.cutoff, dtype=bool)
off_diag_prob = float(np.sum(P_joint_test[~diag_mask]))

print(f"Twin-Photon Fock Correlation Check (r = {r_test}):")
print(f"- Total Off-Diagonal Probability P(na != nb): {off_diag_prob:.2e} (Expected: 0.0)")
assert off_diag_prob < 1e-15, f"Off-diagonal correlations detected: {off_diag_prob:.2e} >= 1e-15!"

# -------------------------------------------------------------
# Part 2: DGCZ Variance Sum Benchmark (Challenge 2)
# -------------------------------------------------------------
r_dgcz = 1.0
psi_dgcz = tmsv_engine.state_tmsv(r_dgcz)
var_sum_emp, vx_emp, vp_emp = tmsv_engine.dgcz_variance_sum(psi_dgcz)
expected_dgcz = np.exp(-2.0 * r_dgcz)  # exp(-2) = 0.1353352832366127

rel_err_dgcz = np.abs(var_sum_emp - expected_dgcz) / expected_dgcz
print(f"\nDGCZ Inseparability Benchmark (r = {r_dgcz}):")
print(f"- Empirical Variance Sum:  {var_sum_emp:.6f}")
print(f"- Analytical e^(-2r):     {expected_dgcz:.6f}")
print(f"- Relative Error:         {rel_err_dgcz:.2e}")

assert rel_err_dgcz < 1e-3, f"DGCZ variance sum deviates by > 0.1%: err = {rel_err_dgcz:.2e}!"
assert var_sum_emp < 1.0, "DGCZ criterion failed to witness entanglement (< 1.0)!"

# -------------------------------------------------------------
# Part 3: Reduced State Thermal Distribution (Challenge 3)
# -------------------------------------------------------------
r_therm = 0.75
psi_therm = tmsv_engine.state_tmsv(r_therm)
P_joint_therm = tmsv_engine.joint_photon_distribution(psi_therm)
p_marginal_emp = np.sum(P_joint_therm, axis=1)[:9]

n_bar = np.sinh(r_therm)**2
n_indices = np.arange(9)
p_thermal_expected = (1.0 / (1.0 + n_bar)) * (n_bar / (1.0 + n_bar))**n_indices

max_therm_err = np.max(np.abs(p_marginal_emp - p_thermal_expected))
print(f"\nReduced State Thermalization Benchmark (r = {r_therm}, n_bar = {n_bar:.4f}):")
print(f"- Max deviation from Bose-Einstein thermal: {max_therm_err:.2e}")

assert max_therm_err < 1e-10, f"Thermal distribution mismatch: {max_therm_err:.2e} >= 1e-10!"

print("=" * 70)
print("SUCCESS: Two-Mode Squeezing & EPR States Numerically Verified!")
print(f"- Twin-photon correlation: P(na != nb) = {off_diag_prob:.2e} == 0 (PASS)")
print(f"- DGCZ inseparability:     {var_sum_emp:.4f} vs. {expected_dgcz:.4f} (PASS: < 0.1%)")
print(f"- Thermalization:          Max err = {max_therm_err:.2e} (PASS: < 1e-10)")
print("=" * 70)

```

3. **DGCZ Inseparability Witness:** Any non-zero squeezing parameter $r > 0$ strictly violates the classical separability bound $\Delta x_-^2 + \Delta p_+^2 = e^{-2r} < 1.0$, serving as an operational certificate of entanglement.
4. **Entanglement-Induced Thermalization:** Tracing out either subsystem yields a bona fide thermal state $\hat{\rho}_a$ with $\bar{n} = \sinh^2 r$.
